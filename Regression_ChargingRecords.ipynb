```python
# Checking the Multicollinearity among the independent variables

# Select only numeric, independent features for analysis
# Exclude unique IDs (like UserID, ChargerID) and your target variable (like Demand)
features = df[['ChargerCompany', 'ChargerType', 'Duration']]

# Calculate the correlation matrix
corr_matrix = features.corr()
print(corr_matrix)
```

```python
plt.figure(figsize=(8, 6))
# annot=True prints the numbers inside the squares
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", vmin=-1, vmax=1)
plt.title("Correlation Heatmap to Detect Multicollinearity")
plt.show()
```

```python
# 1. Clean data and select features
features = df[['ChargerCompany', 'ChargerType', 'Duration']].dropna()

# 2. Get the correlation matrix
corr_matrix = features.corr()

# 3. Use matrix inversion to find the exact VIF scores
corr_inverse = np.linalg.inv(corr_matrix.values)
vif_values = np.diag(corr_inverse)

# 4. Display results in a clean table
vif_data = pd.DataFrame({
    'Feature': features.columns,
    'VIF': vif_values
})

print(vif_data.sort_values(by="VIF", ascending=False).reset_index(drop=True))
```

```python
df['Log_Demand'] = np.log1p(df['Demand'])
```

```python
df.describe()
```

```python
# Removing -ve values from the dataset

df = df[df["Duration"] >= 0].copy()

# 2. Verify the fix by checking the new minimum value
print("✅ Negative values successfully removed!")
print(f"• New Minimum Duration: {df['Duration'].min()} minutes")
print(f"• Remaining Rows in df: {len(df):,}")
```

```python
df.describe()
```

```python
df.info()
```

```python
df['Duration_days'] = (df['EndDay'] - df['StartDay']) / pd.Timedelta(days=1)
```

```python
df['charging_time'] = (df['EndTime'] - df['StartTime']) / pd.Timedelta(hours=1)
```

```python
df.info()
```

```python
model_df = df[["UserID","ChargerID","ChargerCompany","ChargerType","Duration","Log_Demand","Duration_days","charging_time"]]
```

```python
# input Feature Matrix

X = model_df [["UserID","ChargerID","ChargerCompany","ChargerType","Duration","Log_Demand","Duration_days", "charging_time"]]
X.head()
```

```python
# Target Variable

Y = model_df["Log_Demand"]

Y.head()
```

```python
# splitting data into train and test set

from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.2, random_state=42)

```

```python
from sklearn.preprocessing import StandardScaler


# Initialize StandardScaler
ss = StandardScaler()

# Fit and transform the training data
X_train = ss.fit_transform(X_train)

# Transform the test data
X_test = ss.transform(X_test)

```

Linear Regression

```python
from sklearn.linear_model import LinearRegression
lrg = LinearRegression()

# Training

lrg.fit(X_train, Y_train) 

train_score = lrg.score(X_train, Y_train)

#make predictions

Y_pred = lrg.predict(X_test)

```

```python
from statistics import linear_regression


print('Linear Regression  \n')
print(linear_regression(Y_test,Y_pred))
```

```python
# Prediction on the test set
Y_pred = lrg.predict(X_test)

comparision = pd.DataFrame({"Actual": Y_test.values, "Predicted": Y_pred})

comparision.head()
```

```python
# Model Evaluation
from sklearn.metrics import (mean_squared_error, mean_absolute_error, r2_score)
```

```python
mae = mean_absolute_error(Y_test, Y_pred)
mse = mean_squared_error(Y_test, Y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(Y_test, Y_pred)

print(f"the R2 score is {r2}")
print(f"MSE is {mse}")
print(f"RMSE is {rmse}")
print(f"MAE is {mae}")
```

```python
# adjusted R2
n = X_test.shape[0]
p = X_test.shape[1]

adjusted_r2 = (1 - (1-r2)*(n-1)/(n-p-1))
print(f"The adjusted R2 is {adjusted_r2}")
```

```python
# Training VS Test performance
trained_pred = lrg.predict(X_train)
test_pred = lrg.predict(X_test)

train_r2 = r2_score(Y_train, trained_pred)
test_r2 = r2_score(Y_test, test_pred)

print(f"Trained R2 is {train_r2} and test R2 is {test_r2}")
```

Decision Tree

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Initialize the Decision Tree Regressor

dtr = DecisionTreeRegressor(max_depth=5, random_state=42)

# Fit or train the model using the training data

dtr.fit(X_train, Y_train)

# Make predictions on the test data

Y_pred = dtr.predict(X_test)

# Evaluate Performance

mse = mean_squared_error(Y_test, Y_pred)
r2 = r2_score(Y_test, Y_pred)

print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"R-squared (R²): {r2:.4f}")

```

Random Forest

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
rfr = RandomForestRegressor(n_estimators=100, random_state=42)

rfr.fit(X_train,Y_train)

Y_pred = rfr.predict(X_test)

mse = mean_squared_error(Y_test, Y_pred)
r2 = r2_score(Y_test,Y_pred)

print(f"Mean Squared error (MSE): {mse:.4f}")
print(f"R_Squared (R2 Score):{r2:.4f}")
```

Gradient Boost

```python
from sklearn.ensemble import GradientBoostingRegressor
```

```python
gbr = GradientBoostingRegressor(n_estimators=100, learning_rate=0.1, max_depth=3, random_state=42)

gbr.fit(X_train,Y_train)

Y_pred = gbr.predict(X_test)

rmse = mean_squared_error(Y_test,Y_pred)
r2 = r2_score(Y_test,Y_pred)

print(f"Root Mean Squared error: {rmse:.4f}")
print(f"R2 Score: {r2:.4f}")

```
